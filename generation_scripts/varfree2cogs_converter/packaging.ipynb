```python
import pandas as pd
import os


```

```python
def combine_dfs_in_dir(directory_path, output_name):
    outdf = None
    
    for f in os.listdir(directory_path):
        filepath = os.path.join(directory_path, f)
        if os.path.isfile(filepath):
            newdf = pd.read_csv(filepath, sep='\t', header=None)

            if outdf is None:
                outdf = newdf
            else:
                outdf = pd.concat([outdf, newdf])
        else:
            pass

    if not outdf is None:
        outdf.to_csv(os.path.join(directory_path, output_name) + ".tsv", sep='\t', header=None)
```
